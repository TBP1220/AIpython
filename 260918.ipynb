```python
score=int(input('점수를 입력해주세요'))

if score>=60:
    print('합격입니다.')
else:
    print('불합격입니다.')
print('프로그램 종료')
```

```python
#숫자를 입력받아서 짝수인지 홀수인지 출력하는 프로그램을 작성하시오
number=int(input('숫자를 입력해주세요:'))

if number%2==0:
    print('짝수입니다.')
else:
    print('홀수입니다.')
print('프로그램 종료')
```
