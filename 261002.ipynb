```python
#난수 생성기 4호기(while 반복, 맞힌 횟수 카운팅)
import random

num=random.randint(1,100)
count=0
while True:
    user=int(input('숫자를 입력해주세요 :'))
    #print(num)
    count+=1
    if user==num:
        print('숫자를 맞혔습니다!')
        print(f'숫자를 맞히기에 소모된 횟수는 {count}번입니다.')
        break
    elif user>num:
        print('입력한 숫자가 더 큽니다!')
    else:
        print('입력한 숫자가 더 작습니다!')
```

```python
#asmd 함수 정의
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op = input('연산자를 입력하시오')

def asmd(x,y,op):
    if op=='+':
        print(f'{x}+{y}={x+y}')
    elif op=='-':
        print(f'{x}-{y}={x-y}')
    elif op=='*':
        print(f'{x}*{y}={x*y}')
    elif op=='/':
        print(f'{x}/{y}={x/y:.2f}')
    else:
        print('잘못된 연산자입니다.')

asmd(x,y,op)
```
