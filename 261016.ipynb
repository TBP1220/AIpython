```python
#순차구조, 선택구조 if, elif, else, 반복구조 for, while
#반복구조 명령어 range에서 종료값은 반드시 존재, 시작값을 정하지 않으면 0. 증감값은 마지막.
#(ex)range(2,11,3)
#자료구조 [리스트], {딕셔너리}, (튜플)
#while문에서는 종료값을 while 뒤에, 시작값은 while문을 쓰기 전에, 증감값은 while을 쓰고 난 뒤.
#무한히 반복 = while True, if (조건): break.

#리스트 : 변수 하나에 여러 개의 값 저장. 인덱스 값은 0부터 [입력한 값-1]까지.
```

```python
#리스트 함수 한눈에 보기
#분류	함수/연산	형식	반환값	원본 변경
#추가	append	리스트.append(값)	None	O
#추가	insert	리스트.insert(위치, 값)	None	O
#추가	extend	리스트1.extend(리스트2)	None	O
#삭제	pop	리스트.pop(인덱스) / 리스트.pop()	삭제한 항목	O
#삭제	del (키워드)	del 리스트[인덱스]	없음(문장)	O
#삭제	remove	리스트.remove(값)	None	O
#삭제	clear	리스트.clear()	None	O
#재배열	reverse	리스트.reverse()	None	O
#재배열	sort	리스트.sort() / 리스트.sort(reverse=True)	None	O
#재배열	sorted (내장)	sorted(리스트)	정렬된 새 리스트	X
#탐색	index	리스트.index(값)	인덱스(정수)	X
#탐색	in	값 in 리스트	True / False	X
#통계	len	len(리스트)	항목 수(정수)	X
#통계	max	max(리스트)	최댓값	X
#통계	min	min(리스트)	최솟값	X
#통계	sum	sum(리스트)	합계	X
```

```python
#1부터 100까지 합을 출력하시오.
sum=0
for a in range(1, 101):
  sum += a
print(sum)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if 사용)
sum=0
for i in range(1,101):
  if i%2==0:
    sum+=i
print(sum)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if 미사용)
sum=0
for i in range(2,101,2):
  sum += i
print(sum)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요.
x=int(input("단을 입력해주세요."))
for i in range(1,10):
  print(f'{x}x{i}={x*i}')
```

```python
#2단부터 9단까지 구단을 7 x 1 = 7 형식으로 출력하세요.
for i in range(2,10):
  print('-'*20)
  for j in range(1,10):
    print(f'{i}x{j}={i*j}')
```

```python
#while문으로 1부터 100까지 합을 출력하시오.
sum=0
i=1
while i<=100:
  sum+=i
  i+=1
print(sum)
```

```python
#while문으로 1부터 100까지 합을 출력하시오. (break)
sum=0
i=1
while True:
  sum+=i
  i+=1
  if i>100:
    break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요. (while True와 break 사용)
sum=0
while True:
  n=int(input("정수를 입력해주세요."))
  sum+=n
  if n==0:
    break
print(f'입력하신 숫자의 합계는 {sum}입니다.')
```

```python
prices = [1500, 3200, 2500, 4000, 5500]

for i in range(len(prices)):
    prices[i] = prices[i] + prices[i]*0.05

print(prices)
```

```python
cubes = []
for x in range(1, 11):
    cubes.append(x**3)
print(cubes)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 만들어 출력하세요. + 각종 명령어 실험
evens=[]
for i in range(1,21):
  if i%2==0:
    evens.append(i)
print(evens)
print(evens[-1])
print(evens[1:3])
evens.pop(7)
print(evens)
del evens[5]
print(evens)
evens.remove(2)
print(evens)
evens.clear()
print(evens)
```

```python
#함수의 중요성
scores = [85, 95, 70, 75, 99, 93, 88]
print(sum(scores))
```

```python
#수작업 1
total=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  total+=i
print(total)
```

```python
#수작업 2
max_score=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  if max_score<i:
    max_score=i
print(max_score)
```
