```python
#딕셔너리 접근
prices = {'아메리카노':2000,'라떼':3000,'녹차':2500}

for key in prices:
    print(key)

for key in prices:
    print(prices[key])

for key, value in prices.items():
    print(key, value)
```

```python
#딕셔너리 항목 삭제(pop, del, clear)
prices = {'아메리카노':2000,'라떼':3000,'녹차':2500}
a = prices.pop('라떼')
print(a)
print(prices)

prices = {'아메리카노':2000,'라떼':3000,'녹차':2500}
del prices['라떼']
print(prices)

prices = {'아메리카노':2000,'라떼':3000,'녹차':2500}
prices.clear()
print(prices)
```

```python
#실습
prices = {'아메리카노':2000,'라떼':3000,'녹차':2500}
sum = 0

while True:
    s = input('음료 종류 또는 종료: ')
    if s=='종료':
        break
    elif s in prices:
        n = int(input(f'{s} 주문 수: '))
        sum += prices[s]*n
    else:
        print(f'{s}는 없습니다.')

print(f'총 금액: {sum}원')
```

```python
#리스트 내포=담을 값이 먼저, 조건이 뒤. 걸러내는 게 필요하다면 if문을 맨 뒤에 붙이고, else에 따라 걸러내야한다면 앞으로 붙임.
sequence = []
for x in range(1, 101):
    if x % 7 == 0:
        sequence.append(x)
print(sequence)

sequence = [x for x in range(1, 101) if x % 7 == 0]
print(sequence)
```

```python
score = [71, 80, 60, 90, 65]
result = []

for a in score:
    if a >= 70:
        result.append('합격')
    else:
        result.append('불합격')
print(result)

score = [71, 80, 60, 90, 65]
result = ['합격' if a >= 70 else '불합격' for a in score]
print(result)
```

```python
score_result=dict(zip(score, result))
print(score_result)
```
