```python
# 실습 1. 배열 만들기와 배열 정보 보기
import numpy as np                      # [NumPy] numpy를 불러오고 np라는 짧은 이름을 붙인다

print('-- (1) 리스트 -> 배열 --')
scores = [80, 90, 75]                   # 파이썬 리스트
arr = np.array(scores)                  # [NumPy] np.array(): 리스트를 배열(ndarray)로 바꾼다
print(scores, type(scores))
print(arr, type(arr))                   # 배열은 쉼표 없이 출력된다

print('-- (2) 배열의 정보 --')
print('dtype:', arr.dtype)              # [NumPy] dtype: 원소의 자료형 (int64 등)
print('ndim :', arr.ndim)               # [NumPy] ndim: 차원 수 (1차원이면 1)
print('shape:', arr.shape)              # [NumPy] shape: 모양, 1차원은 (원소개수,)
print('size :', arr.size)               # [NumPy] size: 전체 원소 개수

print('-- (3) 2차원 배열 --')
table = np.array([[1, 2, 3], [4, 5, 6]])   # [NumPy] 리스트 안의 리스트 -> 2차원 배열
print(table)
print('ndim :', table.ndim)
print('shape:', table.shape)            # (2, 3) = 2행 3열
print('size :', table.size)

print('-- (4) 자료형 --')
mixed = np.array([1, 2.5, 3])           # 정수와 실수가 섞이면 모두 실수(float)가 된다
print(mixed, mixed.dtype)

print('-- (5) astype 으로 자료형 바꾸기 --')
f = np.array([1.9, 2.7, 3.5])
i = f.astype(int)                       # [NumPy] astype(): 자료형 변환, 소수점 아래는 버려진다
print(f, '->', i)
```

```python
# 실습 2. 배열 만드는 함수들
import numpy as np                      # [NumPy] 라이브러리 불러오기

print('-- (1) zeros / ones / full --')
print(np.zeros(5))                      # [NumPy] zeros(): 0으로 채운 1차원 배열 (실수 0.)
print(np.ones((2, 3)))                  # [NumPy] ones(): 1로 채운 2행 3열 (모양은 튜플로)
print(np.full((2, 2), 7))               # [NumPy] full(): 지정한 값(7)으로 채운 배열
print(np.eye(3))                        # [NumPy] eye(): 대각선이 1인 단위행렬

print('-- (2) arange --')
print(np.arange(5))                     # [NumPy] arange(끝): 0부터 끝-1 까지
print(np.arange(2, 10))                 # arange(시작, 끝): 끝 값은 포함되지 않는다
print(np.arange(0, 10, 2))              # arange(시작, 끝, 간격)
print(np.arange(0, 1, 0.25))            # 간격은 실수도 가능

print('-- (3) linspace --')
print(np.linspace(0, 1, 5))             # [NumPy] linspace(시작, 끝, 개수): 끝 값 포함, 균등 분할
print(np.linspace(0, 100, 11))          # 0부터 100까지 11개 -> 10씩 증가

print('-- (4) arange 와 linspace 비교 --')
print(np.arange(0, 1, 0.25))            # 간격(0.25)을 정한다 -> 1은 빠진다
print(np.linspace(0, 1, 5))             # 개수(5)를 정한다   -> 1이 들어간다
```

```python
# 실습 3. 1차원 배열의 인덱싱과 슬라이싱
import numpy as np                      # [NumPy] 라이브러리 불러오기

arr = np.array([10, 20, 30, 40, 50])    # [NumPy] 1차원 배열 만들기
print('arr =', arr)

print('-- (1) 인덱싱 --')
print(arr[0])                           # [NumPy] arr[i]: 첫 번째 원소 (0부터 시작)
print(arr[-1])                          # 음수 인덱스: 맨 뒤에서 첫 번째

print('-- (2) 슬라이싱 --')
print(arr[1:4])                         # [NumPy] arr[a:b]: 1~3번 인덱스 (4는 제외)
print(arr[:3])                          # 처음부터 2번 인덱스까지
print(arr[::2])                         # [NumPy] arr[::2]: 2칸씩 건너뛰기
print(arr[::-1])                        # 순서 뒤집기

print('-- (3) 값 바꾸기 --')
arr[0] = 99                             # 인덱스에 값을 대입하면 원소가 바뀐다
print(arr)

print('-- (4) 여러 원소 한 번에 선택 --')
print(arr[[0, 2, 4]])                   # [NumPy] 인덱스 리스트: 0, 2, 4번 원소만 뽑는다

print('-- (5) 슬라이스는 원본을 공유한다 --')
part = arr[1:3]                         # [NumPy] 슬라이스는 복사가 아니라 "뷰"(원본과 연결)
part[0] = 0
print('part =', part)
print('arr  =', arr)                    # 원본의 1번 원소도 0으로 바뀌었다

safe = arr[1:3].copy()                  # [NumPy] copy(): 원본과 분리된 복사본
safe[0] = -1
print('safe =', safe, ' arr =', arr)    # 원본은 그대로
```

```python
# 실습 5. 배열 연산
import numpy as np                      # [NumPy] 라이브러리 불러오기

print('-- (1) 리스트의 + 와 * --')
x = [1, 2, 3]
y = [4, 5, 6]
print(x + y)                            # 리스트의 +는 이어 붙이기
print(x * 2)                            # 리스트의 *는 반복

print('-- (2) 배열 + 숫자 --')
a = np.array([1, 2, 3])                 # [NumPy] 배열 만들기
print(a + 10)                           # [NumPy] 모든 원소에 10을 더한다 (원소별 연산)
print(a * 2)                            # 모든 원소에 2를 곱한다
print(a ** 2)                           # 모든 원소를 제곱

print('-- (3) 배열 + 배열 --')
b = np.array([4, 5, 6])
print(a + b)                            # [NumPy] 같은 위치의 원소끼리 더한다
print(a * b)                            # 같은 위치끼리 곱한다
print(b / a)                            # 같은 위치끼리 나눈다 (결과는 실수)

print('-- (4) 수학 함수와 비교 --')
print(np.sqrt(np.array([1, 4, 9, 16]))) # [NumPy] sqrt(): 모든 원소의 제곱근
print(a > 2)                            # [NumPy] 비교 연산 -> True/False 배열

print('-- (5) 브로드캐스팅 --')
m = np.array([[1, 2, 3], [4, 5, 6]])    # 2행 3열
row = np.array([10, 20, 30])            # 길이 3인 1차원 배열
print(m + row)                          # [NumPy] 브로드캐스팅: row가 각 행에 자동으로 더해진다

print('-- (6) 크기가 안 맞으면? --')
try:
    print(np.array([1, 2, 3]) + np.array([1, 2]))   # 길이가 다른 배열끼리의 연산
except ValueError as e:
    print('오류 발생:', e)              # [NumPy] 모양이 맞지 않으면 ValueError
```

```python
# 실습 6. 배열의 모양 바꾸기
import numpy as np                      # [NumPy] 라이브러리 불러오기

print('-- (1) reshape --')
a = np.arange(12)                       # [NumPy] 0~11 (1차원, 원소 12개)
print(a)
m = a.reshape(3, 4)                     # [NumPy] reshape(행, 열): 3행 4열로 모양 변경
print(m)
print('shape:', m.shape)
print('a는 그대로:', a.shape)           # reshape는 새 배열을 돌려줄 뿐 a는 바뀌지 않는다

print('-- (2) -1 사용 --')
print(a.reshape(2, -1).shape)           # [NumPy] -1: 나머지 크기를 자동 계산 -> (2, 6)
print(a.reshape(-1, 3).shape)           # -> (4, 3)

print('-- (3) 전치 --')
print(m.T)                              # [NumPy] .T: 전치(행과 열 맞바꾸기)
print('shape:', m.T.shape)

print('-- (4) 펼치기 --')
print(m.flatten())                      # [NumPy] flatten(): 1차원으로 펼친 복사본

print('-- (5) 합치기 --')
p = np.array([1, 2, 3])
q = np.array([4, 5, 6])
print(np.vstack([p, q]))                # [NumPy] vstack(): 위아래로 쌓기 -> 2행 3열
print(np.hstack([p, q]))                # [NumPy] hstack(): 좌우로 붙이기 -> 길이 6

print('-- (6) 원소 수가 안 맞으면? --')
try:
    np.arange(6).reshape(4, 2)          # 6개를 8칸(4x2)으로는 만들 수 없다
except ValueError as e:
    print('오류 발생:', e)
```

```python
# 실습 7. 통계 함수와 axis
import numpy as np                      # [NumPy] 라이브러리 불러오기

score = np.array([72, 85, 90, 64, 88])  # 한 학생의 5과목 점수
print('score =', score)

print('-- (1) 기본 통계 --')
print('합계  :', np.sum(score))         # [NumPy] sum(): 합계
print('평균  :', np.mean(score))        # [NumPy] mean(): 평균
print('최대  :', np.max(score))         # [NumPy] max(): 최댓값
print('최소  :', np.min(score))         # [NumPy] min(): 최솟값
print('표준편차:', round(np.std(score), 2))   # [NumPy] std(): 표준편차

print('-- (2) 위치 구하기 --')
print('최댓값 위치:', np.argmax(score))  # [NumPy] argmax(): 최댓값이 있는 인덱스
print('최솟값 위치:', np.argmin(score))  # [NumPy] argmin(): 최솟값이 있는 인덱스

print('-- (3) 2차원 배열과 axis --')
table = np.array([[80, 70, 90],          # 학생 A
                  [60, 85, 75],          # 학생 B
                  [95, 90, 100],         # 학생 C
                  [50, 65, 70]])         # 학생 D  (열: 국어, 영어, 수학)
print(table.shape)                       # (4, 3) = 4명 x 3과목
print('전체 평균  :', np.mean(table))
print('과목별 평균:', np.mean(table, axis=0))   # [NumPy] axis=0: 세로 방향 -> 열(과목)마다 계산
print('학생별 평균:', np.mean(table, axis=1))   # [NumPy] axis=1: 가로 방향 -> 행(학생)마다 계산
print('학생별 합계:', np.sum(table, axis=1))
print('과목별 최고:', np.max(table, axis=0))

print('-- (4) 정렬과 누적합 --')
a = np.array([5, 2, 9, 1])
print(np.sort(a))                        # [NumPy] sort(): 오름차순 정렬 (원본은 그대로)
print(np.cumsum(a))                      # [NumPy] cumsum(): 누적합 -> 5, 7, 16, 17
```

```python
# 실습 8. 조건으로 골라내기
import numpy as np                      # [NumPy] 라이브러리 불러오기

score = np.array([55, 82, 67, 90, 45, 73])
print('score =', score)

print('-- (1) 조건 -> 마스크 --')
mask = score >= 70                      # [NumPy] 비교 결과는 True/False 배열(마스크)
print(mask)

print('-- (2) 불리언 인덱싱 --')
print(score[mask])                      # [NumPy] 마스크가 True인 위치의 값만 선택
print(score[score < 60])                # 조건을 바로 써도 된다

print('-- (3) 개수와 비율 --')
print('70점 이상 인원:', np.sum(score >= 70))     # [NumPy] True는 1로 계산 -> 개수
print('70점 이상 비율:', np.mean(score >= 70))    # 평균 = 비율

print('-- (4) np.where --')
result = np.where(score >= 70, '합격', '불합격')    # [NumPy] where(조건, 참일 때 값, 거짓일 때 값)
print(result)
curved = np.where(score < 60, 60, score)           # 60점 미만은 60점으로 올려준다 (최저 60점)
print(curved)

print('-- (5) 조건 결합 --')
mid = score[(score >= 60) & (score < 80)]          # [NumPy] & : 그리고 (각 조건에 괄호)
print('60~79점:', mid)
edge = score[(score < 50) | (score >= 90)]         # [NumPy] | : 또는
print('50 미만 또는 90 이상:', edge)

print('-- (6) any / all --')
print(np.any(score >= 90))              # [NumPy] any(): 하나라도 True 인가?
print(np.all(score >= 60))              # [NumPy] all(): 모두 True 인가?
```

```python
# 실습 10. 종합 실습: 성적 데이터 분석
import numpy as np                      # [NumPy] 라이브러리 불러오기

print('-- (1) 데이터 준비 --')
names = np.array(['민수', '지연', '철수', '영희', '준호'])    # [NumPy] 이름 배열 (문자열도 가능)
subjects = ['국어', '영어', '수학']
scores = np.array([[85, 90, 78],
                   [92, 88, 95],
                   [58, 72, 64],
                   [76, 55, 81],
                   [88, 94, 90]])       # [NumPy] 5명 x 3과목 2차원 배열
print('shape:', scores.shape)           # [NumPy] shape: (5, 3)

print('-- (2) 과목별 분석 (axis=0) --')
print('과목별 평균:', np.round(np.mean(scores, axis=0), 1))   # [NumPy] axis=0: 과목(열)마다
print('과목별 최고:', np.max(scores, axis=0))
print('과목별 최저:', np.min(scores, axis=0))
print('수학 점수만:', scores[:, 2])      # [NumPy] m[:, 열]: 한 열 전체

print('-- (3) 학생별 분석 (axis=1) --')
total = np.sum(scores, axis=1)          # [NumPy] axis=1: 학생(행)마다 합계
avg = np.round(np.mean(scores, axis=1), 1)   # [NumPy] 학생별 평균
for i in range(len(names)):
    print(names[i], '총점', total[i], '평균', avg[i])
best = np.argmax(total)                 # [NumPy] argmax(): 총점이 가장 큰 학생의 위치
print('1등:', names[best], total[best], '점')

print('-- (4) 조건 분석 --')
print('전체 평균:', np.round(np.mean(scores), 1))
print('평균 이상 학생:', names[avg >= np.mean(scores)])      # [NumPy] 불리언 인덱싱
print('과목별 60점 미만 인원:', np.sum(scores < 60, axis=0))   # [NumPy] 조건 개수 + axis

print('-- (5) 학점 매기기 --')
grade = np.where(avg >= 90, 'A',
         np.where(avg >= 80, 'B',
          np.where(avg >= 70, 'C', 'D')))     # [NumPy] where를 겹쳐서 여러 등급 만들기
for n, a, g in zip(names, avg, grade):
    print(n, a, g)
```
